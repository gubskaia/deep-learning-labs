## Задание 1. Ручное применение свёртки

В этом задании нужно выполнить операцию свёртки вручную.

Свёртка — это операция, при которой небольшая матрица, называемая ядром или фильтром, проходит по изображению и вычисляет новые значения.

Дано изображение:

1  2  0
0  1  3
2  1  0

Дано ядро свёртки:

1   0
0  -1

Свёртка выполняется с шагом 1 без padding.

Padding не используется, значит мы не добавляем дополнительные нули по краям изображения.

Размер исходного изображения: 3×3  
Размер ядра: 2×2

Размер итоговой матрицы:

3 - 2 + 1 = 2

Значит, результат будет иметь размер 2×2.

```python
import numpy as np

# Исходное изображение
image = np.array([
    [1, 2, 0],
    [0, 1, 3],
    [2, 1, 0]
])

# Ядро свёртки
kernel = np.array([
    [1, 0],
    [0, -1]
])

print("Исходное изображение:")
print(image)

print("\nЯдро свёртки:")
print(kernel)
```

Теперь выполним свёртку вручную.

Ядро размером 2×2 накладывается на каждый участок изображения размером 2×2.

Первый участок:

1  2
0  1

Вычисление:

1*1 + 2*0 + 0*0 + 1*(-1) = 0

Второй участок:

2  0
1  3

Вычисление:

2*1 + 0*0 + 1*0 + 3*(-1) = -1

Третий участок:

0  1
2  1

Вычисление:

0*1 + 1*0 + 2*0 + 1*(-1) = -1

Четвёртый участок:

1  3
1  0


Вычисление:

1*1 + 3*0 + 1*0 + 0*(-1) = 1

Итоговая матрица признаков:

0  -1
-1  1





```python
# Выполним свёртку с шагом 1 без padding

result = []

# Проходим по строкам изображения
for i in range(image.shape[0] - kernel.shape[0] + 1):
    row = []

    # Проходим по столбцам изображения
    for j in range(image.shape[1] - kernel.shape[1] + 1):

        # Берём участок изображения размером 2×2
        region = image[i:i+2, j:j+2]

        # Умножаем участок на ядро и складываем значения
        value = np.sum(region * kernel)

        row.append(value)

    result.append(row)

result = np.array(result)

print("Итоговая матрица признаков:")
print(result)
```

### Вывод по заданию 1

Получилась итоговая матрица признаков:

0  -1
-1  1

Данное ядро сравнивает левый верхний пиксель с правым нижним пикселем.

Если значения сильно отличаются, результат получается не равным нулю.

Поэтому такое ядро может находить диагональные изменения яркости на изображении.

## Задание 2. Сравнение двух ядер

Даны два ядра.

Ядро A:

1  1
1  1

Ядро B:

1  -1
1  -1

Нужно объяснить, чем они отличаются и какое лучше подходит для выделения границ.



```python
kernel_A = np.array([
    [1, 1],
    [1, 1]
])

kernel_B = np.array([
    [1, -1],
    [1, -1]
])

print("Ядро A:")
print(kernel_A)

print("\nЯдро B:")
print(kernel_B)
```

### Объяснение ядра A

Ядро A состоит только из положительных значений.

Такое ядро складывает значения пикселей в выбранной области.

Если применить его к изображению, оно будет усиливать общую яркость участка.

По смыслу оно похоже на сглаживающий или суммирующий фильтр.

Такое ядро не очень хорошо выделяет границы, потому что в нём нет отрицательных значений.

### Объяснение ядра B

Ядро B содержит положительные и отрицательные значения.

Левая часть ядра положительная, а правая отрицательная.

Это значит, что ядро сравнивает левую и правую части изображения.

Если слева и справа значения пикселей сильно отличаются, результат будет большим по модулю.

Поэтому такое ядро лучше подходит для выделения границ.

Особенно хорошо оно может находить вертикальные границы, потому что сравнивает левую и правую стороны.

```python
# Проверим работу двух ядер на том же изображении

def convolution_2x2(image, kernel):
    result = []

    for i in range(image.shape[0] - kernel.shape[0] + 1):
        row = []

        for j in range(image.shape[1] - kernel.shape[1] + 1):
            region = image[i:i+2, j:j+2]
            value = np.sum(region * kernel)
            row.append(value)

        result.append(row)

    return np.array(result)


result_A = convolution_2x2(image, kernel_A)
result_B = convolution_2x2(image, kernel_B)

print("Результат после применения ядра A:")
print(result_A)

print("\nРезультат после применения ядра B:")
print(result_B)
```

### Вывод по заданию 2

Ядро A складывает значения пикселей и больше подходит для получения общей информации о яркости участка.

Ядро B сравнивает левую и правую части изображения.

Для выделения границ лучше подходит ядро B, потому что оно содержит положительные и отрицательные значения.

После применения ядра A изображение будет больше похоже на сглаженное или усиленное по яркости.

После применения ядра B будут выделяться места, где резко меняется яркость.

## Задание 3. Pooling

Pooling — это операция уменьшения размера карты признаков.

В этом задании нужно выполнить:

* max pooling;
* average pooling.

Дана матрица признаков:

1  3  2  4
5  6  1  2
7  2  8  1
0  3  4  5

Окно pooling имеет размер 2×2.

Шаг равен 2.

Это значит, что мы берём блоки 2×2 без пересечения.

```python
feature_map = np.array([
    [1, 3, 2, 4],
    [5, 6, 1, 2],
    [7, 2, 8, 1],
    [0, 3, 4, 5]
])

print("Исходная матрица признаков:")
print(feature_map)
```

### Max pooling

Max pooling берёт максимальное значение из каждого окна 2×2.

Первый блок:

1  3
5  6

Максимум:

Второй блок:

2  4
1  2

Максимум:

4

Третий блок:

7  2
0  3

Максимум:

7

Четвёртый блок:

8  1
4  5

Максимум:

Итог max pooling:

6  4
7  8








```python
max_pool_result = []

for i in range(0, feature_map.shape[0], 2):
    row = []

    for j in range(0, feature_map.shape[1], 2):
        region = feature_map[i:i+2, j:j+2]
        max_value = np.max(region)
        row.append(max_value)

    max_pool_result.append(row)

max_pool_result = np.array(max_pool_result)

print("Результат Max Pooling:")
print(max_pool_result)
```

### Average pooling

Average pooling берёт среднее значение из каждого окна 2×2.

Первый блок:

1  3
5  6

Среднее значение:

(1 + 3 + 5 + 6) / 4 = 3.75

Второй блок:

2  4
1  2

Среднее значение:

(2 + 4 + 1 + 2) / 4 = 2.25

Третий блок:

7  2
0  3

Среднее значение:

Четвёртый блок:

8  1
4  5

Среднее значение:

(8 + 1 + 4 + 5) / 4 = 4.5

Итог average pooling:

3.75  2.25
3.00  4.50






```python
avg_pool_result = []

for i in range(0, feature_map.shape[0], 2):
    row = []

    for j in range(0, feature_map.shape[1], 2):
        region = feature_map[i:i+2, j:j+2]
        avg_value = np.mean(region)
        row.append(avg_value)

    avg_pool_result.append(row)

avg_pool_result = np.array(avg_pool_result)

print("Результат Average Pooling:")
print(avg_pool_result)
```

### Разница между max pooling и average pooling

Max pooling выбирает самое большое значение из области.

Он сохраняет самый сильный признак.

Например, если в области есть ярко выраженная граница или важная деталь, max pooling оставит её.

Average pooling считает среднее значение.

Он делает карту признаков более сглаженной.

То есть max pooling лучше сохраняет сильные признаки, а average pooling усредняет информацию.

### Зачем pooling уменьшает размер карты признаков?

Pooling уменьшает размер карты признаков по нескольким причинам:

* уменьшается количество вычислений;
* сеть работает быстрее;
* уменьшается количество параметров;
* снижается риск переобучения;
* остаются только самые важные признаки.

Например, матрица 4×4 после pooling с окном 2×2 и шагом 2 превращается в матрицу 2×2.

## Задание 4. Логика работы CNN

CNN - это сверточная нейронная сеть.

Она часто используется для обработки изображений.

Главная идея CNN заключается в том, что сеть сначала извлекает признаки изображения, а потом выполняет классификацию.



### 1. Почему в CNN сначала применяются свёртки, а не сразу классификация?

Изображение состоит из большого количества пикселей.

Если сразу передать все пиксели в обычную полносвязную сеть, модель будет иметь очень много параметров.

Также обычная сеть плохо понимает расположение пикселей рядом друг с другом.

Свёртки нужны для того, чтобы сначала найти полезные признаки изображения:

* линии;
* края;
* углы;
* текстуры;
* формы.

После того как признаки найдены, классифицировать изображение становится проще.

Например, если сеть распознаёт цифру, то сначала она может найти линии и изгибы, а потом понять, какая это цифра.

### 2. Зачем используются несколько фильтров?

Один фильтр ищет только один тип признака.

Например:

* один фильтр может искать вертикальные линии;
* другой фильтр может искать горизонтальные линии;
* третий фильтр может искать диагональные линии;
* четвёртый фильтр может искать углы.

Поэтому в CNN используется несколько фильтров.

Каждый фильтр создаёт свою карту признаков.

Чем больше фильтров, тем больше разных признаков может найти сеть.

### 3. Почему после свёртки часто применяют pooling?

После свёртки получается карта признаков.

Она может быть достаточно большой.

Pooling применяется для уменьшения размера этой карты.

Это нужно, чтобы:

* уменьшить количество вычислений;
* ускорить обучение;
* оставить наиболее важные признаки;
* сделать модель менее чувствительной к небольшим сдвигам изображения.

Например, если объект на изображении немного сдвинулся, pooling помогает сети всё равно распознать его.

### 4. Чем CNN лучше обычной полносвязной сети при работе с изображениями?

CNN лучше обычной полносвязной сети для изображений, потому что она учитывает структуру изображения.

В изображениях важны соседние пиксели.

Например, линия или граница появляется именно из-за связи соседних пикселей.

Обычная полносвязная сеть воспринимает пиксели просто как отдельные числа.

CNN работает лучше, потому что:

* учитывает локальные области изображения;
* использует меньше параметров;
* лучше находит визуальные признаки;
* быстрее обучается на изображениях;
* лучше подходит для задач распознавания объектов.

## Задание 5. Анализ архитектуры

Дана архитектура:

Изображение 28×28 → Свёртка → ReLU → MaxPooling → Свёртка → ReLU → MaxPooling

Нужно описать, что происходит на каждом этапе.




### 1. Изображение 28×28

На вход нейронной сети подаётся изображение размером 28×28 пикселей.

Например, это может быть изображение цифры.

Каждый пиксель имеет своё значение яркости.

Если изображение чёрно-белое, то каждый пиксель можно представить одним числом.

### 2. Первая свёртка

На этом этапе к изображению применяются фильтры.

Фильтры проходят по изображению и ищут простые признаки.

Первый свёрточный слой обычно ищет:

* края;
* линии;
* углы;
* простые переходы яркости.

Результатом свёртки являются карты признаков.

### 3. ReLU

После свёртки применяется функция активации ReLU.

ReLU расшифровывается как Rectified Linear Unit.

Она работает очень просто:

если значение меньше 0 → заменить на 0
если значение больше 0 → оставить как есть

Например:

-3 → 0
 2 → 2
-1 → 0
 5 → 5

 ReLU нужна для того, чтобы добавить в сеть нелинейность.

Без ReLU сеть хуже обучалась бы и могла бы находить только простые линейные зависимости.



```python
# Пример работы ReLU

values = np.array([-3, 2, -1, 5, 0])

relu_result = np.maximum(0, values)

print("До ReLU:")
print(values)

print("\nПосле ReLU:")
print(relu_result)
```

### 4. Первый MaxPooling

После ReLU применяется MaxPooling.

MaxPooling уменьшает размер карты признаков.

Он берёт максимальное значение из каждого небольшого окна.

Например, если окно 2×2, то из четырёх чисел остаётся только одно — самое большое.

Это помогает сохранить самые сильные признаки и уменьшить размер данных.

### 5. Вторая свёртка

После первого pooling применяется ещё одна свёртка.

Второй свёрточный слой работает уже не с исходным изображением, а с признаками, которые нашёл первый слой.

Поэтому второй слой может находить более сложные признаки.

Например:

* комбинации линий;
* простые формы;
* части объектов;
* текстуры.

### 6. Второй ReLU

После второй свёртки снова применяется ReLU.

Она снова заменяет отрицательные значения на 0.

Это помогает сети лучше обучаться и выделять важные признаки.

### 7. Второй MaxPooling

После второго ReLU снова применяется MaxPooling.

Он ещё раз уменьшает размер карты признаков.

В результате остаётся более компактное представление изображения.

То есть сеть постепенно переходит от большого изображения к небольшому набору важных признаков.

### Почему размер изображения уменьшается после pooling?

Размер уменьшается потому, что pooling заменяет целый блок значений одним числом.

Например, блок 2×2 содержит 4 значения:

1  3
5  6

После max pooling остаётся только одно значение:

6

Поэтому карта признаков становится меньше.

Если была карта 28×28, то после pooling 2×2 с шагом 2 она может стать примерно 14×14.




### Какие признаки извлекает первый и второй слой?

Первый свёрточный слой извлекает простые признаки:

* линии;
* края;
* углы;
* простые изменения яркости.

Второй свёрточный слой извлекает более сложные признаки:

* формы;
* текстуры;
* части объектов;
* комбинации линий и углов.

Таким образом, CNN постепенно переходит от простых признаков к более сложным.
